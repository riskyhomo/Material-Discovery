```python
import pandas as pd

df_predicted = pd.read_csv('/content/predicted_properties.csv')
df_final = pd.read_csv('/content/final_aluminum_data.csv')
```

```python
display(df_predicted.head())
display(df_predicted.info())
display(df_predicted.isnull().sum())
display(df_final.head())
display(df_final.info())
display(df_final.isnull().sum())
```

```python
property_columns = ['Elongation (%)', 'Tensile Strength (MPa)', 'Yield Strength (MPa)', 'strain_hardening_index']

print("Descriptive statistics for predicted properties:")
display(df_predicted[property_columns].describe())

print("\nDescriptive statistics for final aluminum data:")
display(df_final[property_columns].describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

property_columns = ['Elongation (%)', 'Tensile Strength (MPa)', 'Yield Strength (MPa)', 'strain_hardening_index']

for property_col in property_columns:
    # Histograms
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    sns.histplot(df_predicted[property_col], ax=axes[0], kde=True)
    axes[0].set_title(f'Predicted Properties: {property_col}')
    axes[0].set_xlabel(property_col)

    sns.histplot(df_final[property_col].dropna(), ax=axes[1], kde=True)
    axes[1].set_title(f'Final Aluminum Data: {property_col}')
    axes[1].set_xlabel(property_col)

    plt.tight_layout()
    plt.show()

    # Box plots
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    sns.boxplot(y=df_predicted[property_col], ax=axes[0])
    axes[0].set_title(f'Predicted Properties: {property_col}')
    axes[0].set_ylabel(property_col)

    sns.boxplot(y=df_final[property_col].dropna(), ax=axes[1])
    axes[1].set_title(f'Final Aluminum Data: {property_col}')
    axes[1].set_ylabel(property_col)

    plt.tight_layout()
    plt.show()
```

```python
property_columns = ['Elongation (%)', 'Tensile Strength (MPa)', 'Yield Strength (MPa)', 'strain_hardening_index']

# Define what "better" means for each property
# Higher is better for Elongation, Tensile Strength, and Yield Strength
# For strain_hardening_index, a value closer to 0 is better, but for simplicity and
# given the observed distributions in the previous EDA, we'll consider a lower value as "better".
better_is_higher = {
    'Elongation (%)': True,
    'Tensile Strength (MPa)': True,
    'Yield Strength (MPa)': True,
    'strain_hardening_index': False # Lower is better
}

better_compositions = []

# Iterate through each row in df_predicted
for index, predicted_row in df_predicted.iterrows():
    is_better_in_at_least_one = False
    for prop in property_columns:
        predicted_value = predicted_row[prop]
        final_values = df_final[prop].dropna()

        if final_values.empty:
            # If there are no non-null values in df_final for this property,
            # we cannot make a comparison for this property.
            continue

        if better_is_higher[prop]:
            # For properties where higher is better, the predicted value must be
            # greater than the maximum value in df_final.
            if predicted_value > final_values.max():
                is_better_in_at_least_one = True
                break # Found at least one better property, no need to check others for this row
        else:
            # For properties where lower is better, the predicted value must be
            # less than the minimum value in df_final.
            if predicted_value < final_values.min():
                is_better_in_at_least_one = True
                break # Found at least one better property, no need to check others for this row

    if is_better_in_at_least_one:
        better_compositions.append(predicted_row)

df_better_compositions = pd.DataFrame(better_compositions)
display(df_better_compositions)
```

```python
if not df_better_compositions.empty:
    print("Compositions from predicted_properties.csv with strictly better properties:")
    display(df_better_compositions)
else:
    print("No compositions from predicted_properties.csv were found with strictly better properties based on the defined criteria.")
```
